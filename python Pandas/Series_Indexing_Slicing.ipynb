```python
import pandas as pd

marks = {
    "Hindi":88,
    "Maths":78,
    "Eng":87,
    "Sci":48,
    "Math":58,
    "History":79,
}

markSeries = pd.Series(marks)
markSeries
```

```python
# TO edit the given index
markSeries[1] = 100
markSeries
```

```python
# EVven if theres no index with that name , it adds same
markSeries['Geo'] = 90
markSeries
```

```python
# SERIES WITH PYTHON FUNCTIOS

# len/type/dir/sorted/max/min

print(len(markSeries))
print(dir(markSeries))
print(type(markSeries))
print(sorted(markSeries))
print(max(markSeries))
print(min(markSeries))
```

```python
# TYPE CONVERSION:

list(markSeries)

dict(markSeries)
```

```python
# TO SEARCH IF ELEMENT IS PRESENT IN THE SERIES;  (Boolean)

'Hindi' in markSeries  # For index
87 in markSeries.values # for value
```

```python
#  LOOPING 

for i in markSeries:  # for value
    print(i)

for i in markSeries.index:   # for index
    print(i)
```

```python
# ARITHMETIC OPERATOR:

100 - markSeries
```

```python
# RELATIONAL OPERATORS:

50 > markSeries
```

```python
# BOOLEAN INDEXING

markSeries[markSeries>50]  # represents index satisfying the condition only


markSeries[markSeries<70].size
```

```python
# VALUE COUNTS:  number of indexes a value has

#Let we have to count which amount appers more than 1 times

ll = (pd.read_csv('cars.csv'))

new_Series = ll.value_counts()
print(ll)
```
