```python
import pandas as pd

movies = pd.read_csv('imdb-top-1000.csv')
movies
```

```python
genre = movies.groupby('Genre')
genre.sum()
```

```python
# find top 3 Genres by total earning:

movies.groupby('Genre').sum()['Gross'].sort_values(ascending=True).head(3)
```

```python
# find top 3 Genres by total earning:  
movies.groupby('Genre')['Gross'].sum().sort_values(ascending=True).head(3)
```

```python
#  find director with most popularity:  

movies.groupby('Director')['No_of_Votes'].sum().sort_values(ascending=False).head(1)
```

```python
#  find nuber of movies done by each actor:

movies.groupby('Star1')['Series_Title'].count().sort_values(ascending=False)
```

```python
# total number of groups

len(movies.groupby('Genre'))
```

```python
# number of items in each group

movies.groupby('Genre').size()
```

```python
#  different agggregate operations for different columns:

genre.sum()

genre.agg({
    'Runtime':'mean',
    'IMDB_Rating':'mean',
    'No_of_Votes':'sum',
    'Gross':'sum',
    'Metascore':'min',
})
```

```python
#  passing ist

genre.agg(['min','max','mean'])
```

```python
# both with...

genre.agg({
    'Runtime':['mean','min'],
    'IMDB_Rating':'mean',
    'No_of_Votes':'sum',
    'Gross':'sum',
    'Metascore':'min',
})
```

```python
#   looping : to find te movie with highest rating in each group

df = pd.DataFrame(columns=movies.columns)  # empty dataframe

#loop:
for group,data in genres:
    df = df.append(data[data['IMDB_Rating'] == data['IMDB_Rating'].max()]) 

df
```

```python
#  another way

df = genres.apply(
    lambda x: x[x['IMDB_Rating'] == x['IMDB_Rating'].max()]
).reset_index(drop=True)

df
```

```python
def rank_movie(genre):
    genre['rank_genre'] = genre['IMDB_Rating'].rank(ascending=False)
    return genre
```

```python
genre.apply(rank_movie)
```

```python
# find normalise imdb ratig group wise:

def normal(group):
    group['norm_rating'] = (group['IMDB_Rating']-group['IMDB_Rating'].min())/(group['IMDB_Rating'].max()-group['IMDB_Rating'].min())
    return group

genre.apply(normal)
```
