```python
import pandas as pd

marks = {
    "Maths": 80,
    "English": 89,
    "Hindi": 85,
    "Science": 60,
    "Math": 50
}
mark = pd.Series(marks)
mark.head(1)
```

```python
mark.head(1)
mark.tail(2)
```

```python
mark.sample(2)
```

```python
mark.sort_index(inplace=True)
mark
# INplace makes permanent changes
```

```python
mark.sort_values(ascending=True).head(1).values[0]
```

```python
mark.sum()
mark.prod()
```

```python
mark.mean()
mark.median()
mark.mode()
mark.std()
mark.var()
```

```python
mark.max()
mark.min()
mark.describe()
```
