```python
import pandas as pd

ipl = pd.read_csv('deliveries.csv')
ipl.head()
```

```python
#  find top 10 batsman in terms of run:

ipl.groupby('batsman')['batsman_runs'].sum().sort_values(ascending=False).head(10)
```

```python
#  find bateman with most number of sixes:

six = ipl[ipl['batsman_runs'] == 6]

six.groupby('batsman')['batsman'].count().sort_values(ascending=False).head(10)
```

```python
#  find batsman with most no of 4,6 in last 5 overs:

temp = ipl[ipl['over'] >15]

temp = temp[(temp['batsman_runs'] == 4) | (temp['batsman_runs'] == 6)]

temp.groupby('batsman')['batsman'].count().sort_values(ascending=False).head(10)

```

```python
#  find v kohli,s record againt all teams

temp = ipl[ipl['batsman'] == 'V Kohli']

temp.groupby('bowling_team')['batsman_runs'].sum().reset_index()
```

```python
#  create a function that can find the highest score of any batsman

def highest(batsman):
    temp = ipl[ipl['batsman'] == batsman]

    return temp.groupby('match_id')['batsman_runs'].sum().sort_values(ascending=False).head(1)

highest('DA Warner')
```
