```python
#drop_duplicates

ser = pd.Series([1,1,12,2,3,3,])

ser.drop_duplicates()
```

```python
# duplicated

ser.duplicated().sum()
```

```python
ser.size  #  counts including null
```

```python
ser.count()   # counts excluding null
```

```python
ser.isnull()  #  checks if nulll is there
```

```python
ser.dropna()  #drops null values
```

```python
ser.fillna(0)  #  edit null values
```

```python
ser[ser.isin([12])]
```
