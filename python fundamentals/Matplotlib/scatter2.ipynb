```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
iris = pd.read_csv('iris.csv')
iris
```

```python
iris['Species'] = iris['Species'].replace({'Iris-setosa':0,'Iris-versicolor':1,'Iris-virginica':2})
iris.sample(5)
```

```python
plt.scatter(iris['SepalWidthCm'],iris['PetalLengthCm'],c=iris['Species'],cmap='jet')
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
plt.colorbar()
plt.figure(figsize=(5,1))
```

```python
x = [4,5,7,4]
y = [1,3,7,9]

plt.scatter(x,y)

plt.text(4,1,'point1')
plt.text(5,3,'point2')
plt.text(7,7,'point3')
plt.text(4,9,'point4')
```

```python
batter = pd.read_csv('batter.csv')

sam = batter.head(100).sample(25,random_state=5)
sam
```

```python
plt.figure(figsize=(16,10))
plt.scatter(sam['avg'],sam['strike_rate'])

plt.axhline(130,color='red')
plt.axvline(30,color='green')

for i in range(sam.shape[0]):
    plt.text(sam['avg'].values[i],sam['strike_rate'].values[i],sam['batter'].values[i],)
```
