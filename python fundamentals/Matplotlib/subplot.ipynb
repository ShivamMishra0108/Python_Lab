```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
bat = pd.read_csv('batter.csv')

fig, ax = plt.subplots(nrows=2,ncols=1,sharex=True)

ax[0].scatter(bat['avg'],bat['runs'])
ax[1].scatter(bat['avg'],bat['strike_rate'],color='orange')

ax[0].set_title('avg vs runs')
ax[0].set_ylabel('runs')

ax[1].set_title('avg vs strikerate')
ax[1].set_ylabel('strikerate')
```
