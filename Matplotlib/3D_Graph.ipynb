```python
fig = plt.figure()

bat = pd.read_csv('batter.csv')
```

```python
ax = plt.subplot(projection='3d')

ax.scatter3D(bat['avg'],bat['runs'],bat['strike_rate'])
ax
```

```python
  #  SURFACE GRAPH
```

```python
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)

xx,yy = np.meshgrid(x,y)

z = xx**2 + yy**2
z.shape

ax = plt.subplot(projection='3d')

p = ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

```python
z = np.cos(xx) + np.cos(yy)
z.shape

ax = plt.subplot(projection='3d')

p = ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

```python
z = np.cos(xx) + np.log(yy)
z.shape

ax = plt.subplot(projection='3d')

p = ax.plot_surface(xx,yy,z,cmap='winter')
fig.colorbar(p)
```
